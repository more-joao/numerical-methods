#### $a) -0.5x^2+2.5x+4.5$

````python
import matplotlib.pyplot as plt
import numpy as np

# definir intervalo a ser percorrido (passo de 0.1)
x = np.arange(-6, 9, 0.1)

# aplicar o método gráfico
h = lambda x: -0.5*(x**2)
g = lambda x: -2.5*(x)-4.5

# plotar as funções h(x) e g(x)
plt.figure(figsize=(4,3))
plt.plot(x, h(x))
plt.plot(x, g(x))
plt.grid()
plt.show()

# definir a função
def f(x):
  return -0.5*(x**2) + 2.5*(x) + 4.5

# nota-se, pelo método gráfico, que as raízes existem nos intervalos:
# [-2.5, 0.0] e [5.0, 7.5]

# construir a função do método:
def bissect(interval=list(), epsilon=float()):
  interval = sorted(interval)
  a = interval[1]
  b = interval[0]
  x = None

  while True:
    if abs(b-a) <= epsilon:
      print(f"|b-a| = {abs(b-a)} <= {epsilon}")
      break
    else:
      x = (a+b)/2
      if f(a)*f(x) < 0:
        b = x
      elif f(b)*f(x) < 0:
        a = x
      print(f"I = {[a,b]} | x = {x} | f(a) = {f(a)} | f(b) = {f(b)} | f(x) = {f(x)} | |b-a| = {abs(b-a)}")
  return x



# i) encontrar raiz no intervalo [-2.5, 0.0]
x = bissect([-2.5, 0.0], 0.01)
print(f"x = {x}")

# de fato, |-1.40512 - (-1.396484375)| = -0.0086 <= 0.01

# ii) encontrar raiz no intervalo [5.0, 7.5]
x = bissect([5.0, 7.5], 0.01)
print(f"x = {x}")

# de fato, |6.40512 -  6.39648| = 0.00864 <= 0.01


````

#### $e) f(x)=3x-e^x$

````python
import matplotlib.pyplot as plt
import numpy as np

# definir intervalo a ser percorrido (passo de 0.1)
x = np.arange(0, 3, 0.1)

# aplicar o método gráfico
h = lambda x: 3*x
g = lambda x: np.e**x

# plotar as funções h(x) e g(x)
plt.figure(figsize=(4,3))
plt.plot(x, h(x))
plt.plot(x, g(x))
plt.grid()
plt.show()

# definir a função
def f(x):
  return 3*x - np.e**x

# nota-se, pelo método gráfico, que as raízes existem nos intervalos:
# [0.0, 1.0] e [1.0, 2.0]

# construir a função do método:
def bissect(interval=list(), epsilon=float()):
  interval = sorted(interval)
  a = interval[1]
  b = interval[0]
  x = None

  while True:
    if abs(b-a) <= epsilon:
      print(f"|b-a| = {abs(b-a)} <= {epsilon}")
      break
    else:
      x = (a+b)/2
      if f(a)*f(x) < 0:
        b = x
      elif f(b)*f(x) < 0:
        a = x
      print(f"I = {[a,b]} | x = {x} | f(a) = {f(a)} | f(b) = {f(b)} | f(x) = {f(x)} | |b-a| = {abs(b-a)}")
  return x

# i) encontrar raiz no intervalo [0.0, 1.0]
x = bissect([0.0, 1.0], 0.01)
print(f"x = {x}")

# ii) encontrar raiz no intervalo [1.0, 2.0]
x = bissect([1.0, 2.0], 0.01)
print(f"x = {x}")




````

#### $j) f(x)=ln(x)-sen(x)$

````python
import matplotlib.pyplot as plt
import numpy as np

# definir intervalo a ser percorrido (passo de 0.1)
x = np.arange(0, 3, 0.1)

# aplicar o método gráfico
h = lambda x: np.log(x)
g = lambda x: np.sin(x)

# plotar as funções h(x) e g(x)
plt.figure(figsize=(4,3))
plt.plot(x, h(x))
plt.plot(x, g(x))
plt.grid()
plt.show()

# definir a função
def f(x):
  return np.log(x)-np.sin(x)

# nota-se, pelo método gráfico, que há uma raiz no intervalo:
# [1.0, 3.0]

# construir a função do método:
def bissect(interval=list(), epsilon=float()):
  interval = sorted(interval)
  a = interval[1]
  b = interval[0]
  x = None

  while True:
    if abs(b-a) <= epsilon:
      print(f"|b-a| = {abs(b-a)} <= {epsilon}")
      break
    else:
      x = (a+b)/2
      if f(a)*f(x) < 0:
        b = x
      elif f(b)*f(x) < 0:
        a = x
      print(f"I = {[a,b]} | x = {x} | f(a) = {f(a)} | f(b) = {f(b)} | f(x) = {f(x)} | |b-a| = {abs(b-a)}")
  return x

# encontrar raiz no intervalo [1.0, 3.0]
x = bissect([1.0, 3.0], 0.01)
print(f"x = {x}")



````
